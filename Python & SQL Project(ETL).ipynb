```python
## read the data and handling the null values

import pandas as pd
df = pd.read_csv('orders.csv',na_values = ['Not Available','unknown'])
```

```python
df.head(20)
```

```python
## Checking whether null values exist
df.isnull().sum()
```

```python
## display the unique values in the column
df['Ship Mode'].unique()
```

```python
## rename column name, make them lower case and replace space with underscore 
## columns converted to string object and lower casing it and then replacing thwe space with underscore
df.columns=df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df.head()
```

```python
## Generating the discounted amount,sale price and profit
df['discount']= df['list_price'] *df['discount_percent'] * 0.01
df['sale_price'] = df['list_price'] - df['discount']
df['profit'] = df['sale_price'] - df['cost_price']
```

```python
df.head()
```

```python
##checking the datatypes and convert the order_date column to datetime dtype
#df.dtypes
df['order_date']= pd.to_datetime(df['order_date'],format=('%Y-%m-%d'))
#df.dtypes
```

```python
## droping the columns and using inplace =true to apply the changes in the same df itself
df.drop(columns = ['list_price','discount_percent','cost_price'],inplace =True)
```
